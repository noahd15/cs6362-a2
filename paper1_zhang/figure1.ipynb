# Figure 1 of Zhang et al. (ICLR 2017) from saved runs

This notebook never trains. It regenerates Figure 1 from the per-run JSON logs in `results_fig1/`, which `run_sweep.sh` / `train_fig1.py` write during training.

Convergence time (panel b) is the first step at which full-training-set accuracy reaches a threshold, relative to the true-label run of the same model; panel (c) reports test error at that step. The main figure uses threshold 0.99; 0.95 and 0.98 are shown as a sensitivity check.

```python
import subprocess, sys
from IPython.display import Image, display

def make_figure(threshold):
    out = subprocess.run([sys.executable, 'plot_fig1.py', '--threshold', str(threshold)],
                         capture_output=True, text=True)
    print(out.stdout[-2000:], out.stderr[-2000:])
    display(Image(f'results_fig1/figure1_thr{threshold}_crossing.png'))
```

## Main figure (threshold 0.99)

```python
make_figure(0.99)
```

## Threshold sensitivity (0.95 and 0.98)

```python
for thr in (0.95, 0.98):
    make_figure(thr)
```
